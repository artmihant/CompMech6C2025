**1. Введение**

Первые работы на тему траектории полёта тела, брошенного под углом к горизонту, были опубликованы в XVII веке. Г. Галилей показал, что такое тело движется по параболической траектории [1]. И. Ньютон в своей работе [2] объяснил причины, по которым движение происходит именно по параболе...

Первые публикации по численным методам появились в XVIII веке. Л. Эйлер в своём трёхтомнике [3] описал простейший численный метод для решения задачи Коши для обыкновенного дифференциального уравнения либо системы таких уравнений. Этот метод впоследствии был назван методом Эйлера. Более точные и универсальные численные методы были описаны в конце XIX - начале XX века немецкими математиками К. Рунге [4] и В. Куттой [5]. Это семейство методов получило название методов Рунге-Кутты...

**2. Математическая постановка задачи о полёте тела, брошенного под углом к горизонту, без учёта сопротивления воздуха**

Полёт тела, брошенного под углом к горизонту, описывается следующей системой обыкновенных дифференциальных уравнений:

$$
\begin{cases}
\frac{dx}{dt} = v_x \\
\frac{dy}{dt} = v_y \\
\frac{dv_x}{dt} = 0 \\
\frac{dv_y}{dt} = -g
\end{cases}
$$

Такая задача имеет аналитическое решение, хорошо известное из школьных учебников:

$$ y(x) = x tg(\alpha) -  \frac{g x^2}{2 v_0^2 cos^2 \alpha}  $$

Изобразим это решение при помощи программы:



```python
import math
import numpy as np
import matplotlib.pyplot as plt

# PHYSICS
v0 = 10
alpha = math.radians(45.0)
g = 9.81
flight_range = v0 * v0 * math.sin(2 * alpha) / g

# NUMERIC
nsteps = 100

# PREPROCESSING
x = np.linspace(0, flight_range, nsteps + 1)
y_an = x * math.tan(alpha) - g * x * x / (2.0 * v0 * v0 * math.cos(alpha) ** 2)

# FIGURE
plt.plot(x, y_an)
plt.axis('equal')
plt.show()
```

**3. Численное решение задачи о полёте тела, брошенного под углом к горизонту, без учёта сопротивления воздуха**

Используем метод Эйлера. Разностная схема для этого метода выглядит так:

$$
\begin{cases}
x^{i+1} = x^i + v_x^i \cdot dt \\ 
y^{i+1} = y^i + v_y^i \cdot dt \\
v_x^{i+1} = v_x^i \\
v_y^{i+1} = v_y^i - g \cdot dt
\end{cases}
$$

Программа, реализующая его:


```python
import math
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as anim

# PHYSICS
v0 = 10
alpha = math.radians(45.0)
g = 9.81

flight_range = v0 * v0 * math.sin(2 * alpha) / g
vx = v0 * math.cos(alpha)
vy = v0 * math.sin(alpha)

# NUMERIC
nsteps = 20
dx = flight_range / nsteps
dt = dx / vx

# PREPROCESSING
x = np.linspace(0, flight_range, nsteps + 1)
y_an = x * math.tan(alpha) - g * x * x / (2.0 * v0 * v0 * math.cos(alpha) ** 2)
y_num = np.zeros(nsteps + 1)

# FIGURE
fig, ax = plt.subplots()
ax.axis('equal')
traject_an = ax.plot(x, y_an, lw=3)[0]
traject_num = ax.plot(x[:1], y_num[:1], lw=3)[0]

def init_anim():
    traject_an.set_data(x, y_an)
    traject_num.set_data(x[:1], y_num[:1])
    return (traject_an, traject_num)

def loop_anim(i):
    global vy
    y_num[i + 1] = y_num[i] + vy * dt
    vy = vy - g * dt
    traject_num.set_data(x[:i+2], y_num[:i+2])
    return (traject_an, traject_num)

ani = anim.FuncAnimation(
    fig=fig, func=loop_anim, init_func=init_anim, 
    frames=nsteps, interval=1, repeat=False
)

plt.show()
```
